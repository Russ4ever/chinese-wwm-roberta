```python
import pandas as pd
D = 'con_sec_fy12'
# 1. 先看表结构（字段名和数据类型）
schema = pd.read_sql(f"""
    SELECT COLUMN_NAME, DATA_TYPE, IS_NULLABLE, COLUMN_DEFAULT, COLUMN_COMMENT
    FROM information_schema.columns 
    WHERE table_name = '{D}'
    ORDER BY ORDINAL_POSITION
""", conn)
print("=== 表结构 ===")
print(schema.to_string(index=False))
```

```python
# 2. 看前几行数据 + 总行数
count = pd.read_sql(f'SELECT COUNT(*) AS cnt FROM {D}', conn).iloc[0]['cnt']
print(f"\n=== 共 {count} 行 ===\n")
```

```python
date_col = "TRADE_DATE"
time_range = pd.read_sql(f"""
    SELECT MIN({date_col}) AS min_date, MAX({date_col}) AS max_date
    FROM {D}
""", conn)
print(f"统计日期范围：{time_range['min_date'].values[0]} → {time_range['max_date'].values[0]}")
print(f"共跨越 {len(pd.date_range(time_range['min_date'].values[0], time_range['max_date'].values[0]))} 个交易日")
```

```python
symbol_col = "TICKER_SYMBOL"
pd.read_sql(f"""
    SELECT 
        AVG(cnt) AS avg_symbols_per_day
    FROM (
        SELECT COUNT(DISTINCT {symbol_col}) AS cnt
        FROM {D}
        WHERE {symbol_col} NOT LIKE '8%'
          AND {symbol_col} NOT LIKE '9%'
        GROUP BY {date_col}
    ) subq
""", conn)
```
